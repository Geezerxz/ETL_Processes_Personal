```python
#import packages

import itertools
from itertools import combinations
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import yfinance as yf
import statsmodels.api as sm
from scipy import stats
from datetime import datetime
from datetime import datetime as dt
from scipy import stats
from scipy.stats import norm
from scipy.stats import exponweib
from scipy.stats import johnsonsu
```

```python
tickers= ['INDA','INDY']

stocklist = []

for i in range(0,len(tickers)):
    # data=yf.download(tickers[i],start_date,end_date,interval='1d')
    data=yf.download(tickers[i],period='max',interval='1d')
    # data=yf.download(tickers[i],period='max',interval='1mo')
    data=data.reset_index(level='Date')
    # data=data.reset_index(level='Datetime')
    data.reset_index(inplace=True,drop=True)
    data=data.droplevel(1, axis=1)
    data['Ticker']=tickers[i]
    data['Return']=(data['Close']-data['Close'].shift(1))/data['Close'].shift(1)
    data['logReturn']=np.log(data['Close']/(data['Close'].shift(1)))
    data=data.dropna()
    stocklist.append(data)
stocklist=pd.concat(stocklist,ignore_index=True)
stocklist.reset_index(inplace=True,drop=True)

print(stocklist)
```
